```python
from pathlib import Path
```

```python
def find_root():
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / 'ISPY2_dataset').is_dir():
            return candidate
    raise FileNotFoundError('ISPY2_dataset introuvable')

PROJECT_ROOT = find_root()
PATIENT = 'ISPY2-125130'
SESSION = 'T1'
RESULTS_DIR = PROJECT_ROOT / 'results' / f'sub-{PATIENT}' / f'ses-{SESSION}'

print(f'Projet : {PROJECT_ROOT}')
print(f'Patient/session : {PATIENT} / {SESSION}')
print(f'Résultats : {RESULTS_DIR}')
```
